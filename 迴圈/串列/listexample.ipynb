# 簡介

串列(list)是一個可以儲存多個元素(整數，浮點數，或字串)的容器

元素都用逗號隔開，左右要有抓括號[]

```python
# sores 是一個串列，裡面每個元素都是整數
scores=[78,29,56,80,99]
print(scores)
```

## 索引直

索引直由0開始

```python
print(scores[0]) # 列印地1個元素
print(scores[1]) # 列印地2個元素
```

## 串列內容的個數

len(a)取得a這個串列中元素的個數

len是length(長度)的意思

```python
print(len(scores))
print(scores[4]) # 取得最後一個元素
print(scores[5-1]) # 取得最後一個元素
print(scores[len(scores)-1])
print(scores[-1]) # 取得最後一個元素
```

```python
print(scores)
scores.append(78)
scores.append(99)
```

```python
print(scores)
scores[2]=70 # 將70指定給索引直為2的元素
```
