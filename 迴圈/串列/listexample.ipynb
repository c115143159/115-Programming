# 簡介

串列(list)是一個可以儲存多個元素(整數，浮點數，或字串)的容器

元素都用逗號隔開，左右要有抓括號[]

```python
# sores 是一個串列，裡面每個元素都是整數
scores=[78,29,56,80,99]
print(scores)
```

## 索引直

索引直由0開始

```python
print(scores[0]) # 列印地1個元素
print(scores[1]) # 列印地2個元素
```

## 串列內容的個數

len(a)取得a這個串列中元素的個數

len是length(長度)的意思

```python
print(len(scores))
print(scores[4]) # 取得最後一個元素
print(scores[5-1]) # 取得最後一個元素
print(scores[len(scores)-1])
print(scores[-1]) # 取得最後一個元素
```

```python
print(scores)
scores.append(78)
scores.append(93)
```

```python
print(scores)
scores[2]=70 # 將70指定給索引直為2的元素
```

```python
weights=[12]*6
print(weights)
```

## 走訪列的3種寫法

```python
print(scores)
for i in range(len(scores)):
    print(scores[i])
```

```python
# 對於 scores 中的每一個元素，一次取一個，放在e中
for e in scores:
    print(e)
```

```python
for i,e in enumerate(scores):
    print(i,e)
```

## 計算費伯納西數列的第91個元素

如下:

1,1,2,3,5,8.....

請問地91個元素為何?

提示:

另$x$是一個費伯納西數列，X[i]是第i個數字

x[n]=x[n-1]+x[n-2]


```python
fib=[0]*91
print(fib)
fib[0]=1
fib[1]=1
print(fib)
i=2
fib[i]=fib[i-2]+fib[i-1]
print(fib)
i=3
fib[i]=fib[i-2]+fib[i-1]
print(fib)
i=4
fib[i]=fib[i-2]+fib[i-1]
print(fib)
```

```python
for i in range(2,91):
    fib[i]=fib[i-2]+fib[i-1]

print(fib[90])
```
