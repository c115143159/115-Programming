# 介紹
本程式使用 pyautogui 來完成一個 Blitz 的遊戲機器人

```python
import pyautogui as pyg
pyg.sleep(2)
basex=65
basey=470
105,470
gap=40
```

```python
# 建立 8x8 寶石盤面各格中心座標，並讀取每格中心的螢幕顏色
rows = 8
cols = 8
board_points = [[(basex + col * gap, basey + row * gap) for col in range(cols)] for row in range(rows)]
board_colors = [[pyg.pixel(x, y) for x, y in row] for row in board_points]

for row in board_colors:
    print(row)
```
