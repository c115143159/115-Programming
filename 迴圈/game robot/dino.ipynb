使用 pyautogui 來完成小恐龍的遊戲機器人

小恐龍是在 chrome 中使用 chromeome://dino 啟動

```python
import pyautogui as pyg
pyg.sleep(2)#sleep 2 sec.
pyg.press("space")

#無窮迴圈
while True:
    im=pyg.screenshot()
    obs=False#假設沒有仙人掌
    for i in range(15):
        c=im.getpixel((143+i,418))
    if c ==(83,83,83):
        obs= True
        break
    if obs == True:
        pyg.press("space")
```
