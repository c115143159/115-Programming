# 九九乘法表

1\*1=1,1\*2=2,1\*3=3,...

2\*1=2,2\*2=2,2\*3=6,...

3\*1=3,3\*2=6,3\*3=9,...

```python
i=1
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end='，')
print()
i=2
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end='，')
print()
i=3
for j in range(1,6):
    print(f'{i}*{j}={i*j}',end='，')
print()
```

```python
for i in range(1,6):
    for j in range(1,6):
       print(f'{i}*{j}={i*j}',end='，')
print() 
```

# 練習印出質數

```python
n=273
i=2

if 273%i==0:
    print(f"{273}不是質數")
if 273%i==0:
    print(f"{273}不是質數")
```

```python
n=273
isPrime = True # 假設 n 是質數
for i in range(2,n):
    if n % i ==0:
        print(f'{n}不是質數，因為它可以被{i}除盡')
        isPrime = False # 確定 n 不是質數
        break # 只要數字能被除盡，就跳過迴圈

if isPrime == True: # 判斷我們的假設是否成立
    print(f'{n}是質數')
```
